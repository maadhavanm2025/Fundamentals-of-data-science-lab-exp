```python
import numpy as np
import pandas as pd

data = {
    'YearsExperience': [1.0, 1.5, 2.2, 2.9, 3.5, 4.0, 4.8, 5.5,
                        6.1, 6.8, 7.5, 8.2, 8.9, 9.5, 10.0],

    'Salary': [30000, 34000, 41000, 47000, 53000, 59000, 65000, 71000,
               77000, 83000, 89000, 95000, 102000, 109000, 115000]
}

df = pd.DataFrame(data)

print(df)
df.info()

df.dropna(inplace=True)

print(df.describe())

features = df[['YearsExperience']].values
label = df[['Salary']].values

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

print("Training Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)

import pickle

pickle.dump(model, open('SalaryPred.model', 'wb'))

model = pickle.load(open('SalaryPred.model', 'rb'))

yr_of_exp = float(input("Enter Years of Experience: "))

salary = model.predict(np.array([[yr_of_exp]]))

print("Predicted Salary:", salary[0][0])
```
