```python
import numpy as np
import pandas as pd
data = {
    "Country": [
        "India", "USA", "UK", "India", "USA",
        "UK", "India", np.nan, "USA", "India"
    ],
    "Age": [
        25, 32, 28, np.nan, 35,
        30, 22, 40, 27, 29
    ],
    "Salary": [
        45000, 65000, np.nan, 50000, 72000,
        58000, 42000, 80000, 60000, 48000
    ],
    "Purchased": [
        "Yes", "No", "Yes", "No", "Yes",
        "No", "Yes", "Yes", "No", "Yes"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nDataset Information:")
df.info()

print("\nMode of Country:")
print(df["Country"].mode())

df["Country"] = df["Country"].fillna(
    df["Country"].mode()[0]
)

df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

df["Salary"] = df["Salary"].fillna(
    round(df["Salary"].mean())
)
print("\nAfter Data Preprocessing:")
print(df)
```
