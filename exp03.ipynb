```python
import pandas as pd
import numpy as np

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "Age Group": [
        "18-25", "26-35", "26-35", "36-45", "18-25",
        "46+", "26-35", "18-25", "36-45", "46+"
    ],

    "Rating(1-5)": [
        4, 6, 3, -2, 5,
        4.5, 7, 2, 3.5, 4
    ],

    "Hotel": [
        "Taj", "Marriott", "Taj", "Hilton", "Marriot",
        "Hilton", "Taj", "Mariott", "Hilton", "Taj"
    ],

    "FoodPreference": [
        "veg", "Non-Veg", "Vegetarian", "non-veg", "Veg",
        "Non-Veg", "veg", "Vegetarian", "non-veg", "Veg"
    ],

    "Bill": [
        1500, 2500, -500, 3200, 1800,
        4500, 2200, -1000, 2800, 3500
    ],

    "NoOfPax": [
        2, 4, 3, 2, -1,
        5, 1, 6, 3, 2
    ],

    "EstimatedSalary": [
        35000, 55000, -20000, 75000, 42000,
        90000, -5000, 68000, 58000, 72000
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

# Replace invalid values with NaN
df.loc[df["Bill"] < 0, "Bill"] = np.nan

df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan

df.loc[df["NoOfPax"] < 1, "NoOfPax"] = np.nan

df.loc[
    (df["Rating(1-5)"] < 1) | (df["Rating(1-5)"] > 5),
    "Rating(1-5)"
] = np.nan

# Correct hotel names
df["Hotel"] = df["Hotel"].replace(
    ["Marriot", "Mariott"], "Marriott"
)

# Standardize FoodPreference
df["FoodPreference"] = df["FoodPreference"].replace(
    ["veg", "Vegetarian"], "Veg"
)

df["FoodPreference"] = df["FoodPreference"].replace(
    ["non-veg", "Non-Veg"], "Non-Veg"
)

# Fill missing values
df["Bill"] = df["Bill"].fillna(
    df["Bill"].mean()
)

df["EstimatedSalary"] = df["EstimatedSalary"].fillna(
    df["EstimatedSalary"].mean()
)

df["NoOfPax"] = df["NoOfPax"].fillna(
    df["NoOfPax"].median()
)

df["Rating(1-5)"] = df["Rating(1-5)"].fillna(
    df["Rating(1-5)"].median()
)

print("\nCleaned Dataset:")
print(df)
```
