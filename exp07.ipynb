```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Different input data
data = {
    'bill': [20, 35, 15, 40, 25, 30, 18, 45],
    'tip': [3, 5, 2, 7, 4, 5, 2, 8],
    'gender': ['M','F','M','F','M','F','M','F'],
    'day': ['Mon','Tue','Wed','Thu','Fri','Sat','Sun','Mon']
}

df = pd.DataFrame(data)

print(df.head())

# Distribution
sns.displot(df['bill'])
plt.show()

# Relationship
sns.jointplot(data=df, x='tip', y='bill', kind='reg')
plt.show()

# Pair plot
sns.pairplot(df, hue='gender')
plt.show()

# Correlation
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()

# Box plot
sns.boxplot(x=df['bill'])
plt.show()

# Count plot
sns.countplot(data=df, x='gender')
plt.show()

# Bar chart
df['gender'].value_counts().plot(kind='bar')
plt.show()
```
