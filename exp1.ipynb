```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Year': list(range(2015, 2026)),
    'Revenue': [120, 150, 180, 220, 280, 350, 430, 520, 610, 750, 900]
}

df = pd.DataFrame(data)

plt.plot(df['Year'], df['Revenue'], marker='o')
plt.title('Company Revenue Growth')
plt.xlabel('Year')
plt.ylabel('Revenue (in Lakhs)')
plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ['HR', 'Finance', 'IT', 'Marketing', 'Sales']
employees = [40, 60, 150, 80, 120]

plt.bar(departments, employees)
plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```

```python
import pandas as pd

# Structured data
structured_data = pd.DataFrame({
    'Roll_No': [101, 102, 103],
    'Student': ['Arun', 'Kavin', 'Rahul'],
    'Marks': [85, 92, 78]
})

print("Structured Data:\n", structured_data)

# Unstructured data
unstructured_data = "Arun scored very well in Python and showed excellent programming skills."

print("\nUnstructured Data:\n", unstructured_data)

# Semi-structured data
semi_structured_data = {
    'Roll_No': 101,
    'Student': 'Arun',
    'Details': {
        'Course': 'Computer Science',
        'Grade': 'A'
    }
}

print("\nSemi-structured Data:\n", semi_structured_data)
```

```python
from cryptography.fernet import Fernet

key = Fernet.generate_key()
cipher_suite = Fernet(key)

plain_text = b"Computer Science Engineering"
cipher_text = cipher_suite.encrypt(plain_text)

# Decrypt data
decrypted_text = cipher_suite.decrypt(cipher_text)

print("Original Data:", plain_text)
print("Encrypted Data:", cipher_text)
print("Decrypted Data:", decrypted_text)
```
